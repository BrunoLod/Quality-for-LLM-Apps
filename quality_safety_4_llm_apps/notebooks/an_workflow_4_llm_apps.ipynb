## Workflow 

De forma geral, o workflow padrão de um desenvolvimento de aplicações baseadas em LLM's devem seguir o seguinte fluxograma. 

O bloco azul se refere à segurança da aplicação segundo a qual visa estabelecer proteção tanto ao usuário quanto ao negócio, impedindo no primeiro caso que informações sensíveis sejam informadas na janela de contexto do modelo de LLM utilizado, enquanto no segundo que o modelo passe a operar segundo a ingerências do usuário, podendo ferir a integridade da aplicação. 

Por outro lado, o bloco verde se circunscreve no monitoramento da qualidade de resposta do modelo, o qual deve ocorrer - ao menos no primeiro momento - junto de uma equipe de curadoria que em função de seu conhecimento a respeito do negócio cria respostas tidas como ground truth a partir das quais o modelo será analisado. Uma vez que as respostas tidas como ground truth estejam estabelecidas, uma análise de forma mais automática pode ocorrer, com base na correlação das métricas quantitativas encontradas durante à análise da resposta do modelo realizada por meio de um teste de hipótese. No arquivo destinado ao tópico que se destina à análise da qualidade de resposta dos modelos, exploro métricas tais como BLEU score, BERT score e Sentence embedding by cossine similarity. Contudo, pode-se ser utilizado o framework [RAGAS](https://docs.ragas.io/en/stable/) para tanto, que ajuda a compreender a qualidade de resposta do modelo em termos de completude e corretude, além de outros valores.
